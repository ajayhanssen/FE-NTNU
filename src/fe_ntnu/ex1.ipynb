```python
import numpy as np
import sympy as sp
```

## Problem 1

### a)

```python
A = np.array([[1, 6], [5, 4], [9, 2]])
B = np.array([[2, 8], [4, 3], [5, 11]])
C = np.array([[-2, 2, 5], [-2, 6, -4]])

print(A+B)
print(A-B)
print(A@C)
print(C@A)
```

### b)

```python
A = np.array([[3/5, -4/5], [4/5, 3/5]])
print(A.T)
print(np.linalg.inv(A))
```

### c)

```python
a = np.array([[1], [2]])
alpha = 60*np.pi/180
R = np.array([[np.cos(alpha), -np.sin(alpha)], [np.sin(alpha), np.cos(alpha)]])
print(R)
print(R@a)
```

### d)

```python
A = np.array([[4, 2], [3, -1]])
b = np.array([[6], [7]])
print(np.linalg.inv(A)@b)
```

### e)

```python
A = np.array([[4, 2], [-2, -1]])
print(np.linalg.det(A))
```

### f)

```python
A = np.array([[2, -1, 0], [1, -3, 1], [-1, 1, -3]])
b = np.array([[2], [-2], [-6]])
print(np.linalg.solve(A,b))
```
