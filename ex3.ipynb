```python
import sympy as sp
import numpy as np
```

## Problem 1

```python
E, I, a, p = sp.symbols('E, I, a, p')
L = 2*a
r = 5*p*a**4/(16*E*I)

k4 = 2*E*I/L**3*sp.Matrix([[12, -6*L, -12, -6*L],
                [-6*L, 4*L**2, 6*L, 2*L**2],
                [-12, 6*L, 12, 6*L],
                [-6*L, 2*L**2, 6*L, 4*L**2]])

k3 = 3*E*I/L**3*sp.Matrix([[1, -L, -1],
                           [-L, L**2, L],
                           [-1, L, 1]])

a1 = sp.Matrix([[0], [0], [1], [0]])
a2 = sp.Matrix([[1], [0], [0]])
```

```python
S1 = k4*a1*r
print(S1)
```

```python
So2 = p*L/8*sp.Matrix([[-5], [L], [-3]])

S2 = k3*a2*r+So2
S2
```

## Problem 2

```python
alpha = np.atan2(3, 4)
print(alpha)
T_0 = np.array([[np.cos(alpha), np.sin(alpha)],
                [-np.sin(alpha), np.cos(alpha)]])
T_g = np.zeros((4,4))
T_g[0:2, 0:2] = T_0 
T_g[2:4, 2:4] = T_0
print(T_g)

v_gl = np.array([1, 0.5, 1, 0.5]).T

v_8_6 = T_g @ v_gl
print(v_8_6)
```
