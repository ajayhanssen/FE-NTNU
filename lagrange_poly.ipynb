```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
zeta = np.linspace(-1, 1, 201)
eta = np.linspace(-1, 1, 201)

zzeta, eeta = np.meshgrid(zeta, eta)

N1 = 1/4*(1-zzeta)*(1-eeta)
N2 = 1/4*(1+zzeta)*(1-eeta)
N3 = 1/4*(1+zzeta)*(1+eeta)
N4 = 1/4*(1-zzeta)*(1+eeta)

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(zzeta, eeta, N1, vmin=N1.min() * 2)
ax.plot_surface(zzeta, eeta, N2, vmin=N1.min() * 2)
ax.plot_surface(zzeta, eeta, N3, vmin=N1.min() * 2)
#ax.plot_surface(zzeta, eeta, N4, vmin=N1.min() * 2)

ax.plot_surface(zzeta, eeta, N1+N2+N3+N4, vmin=N1.min() * 2)


print(zzeta)
```
