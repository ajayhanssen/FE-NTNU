## c)

```python
L, E, A = sp.symbols('L E A')

A_ = sp.Matrix([[L*sp.Rational(1,2), L**2*sp.Rational(1,4)], [L, L**2]])
k_q = sp.Matrix([[E*A*L, E*A*L**2], [E*A*L**2, sp.Rational(4,3)*E*A*L**3]])

A_inv = A_.inv()
A_inv.T @ k_q @ A_inv
```

## d)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
L = 2       # m
p0 = 1000   # N/m
E = 210000000 # Pa
A = 1e-4

x = np.linspace(0,L,201)

u_q = -3*p0*L**2/(20*E*A)*x**2 + 2*p0*L**3/(5*E*A)*x
u_L = p0*L**2/(20*E*A)*(-3*x**2+8*L*x)
u_analytic = p0/(12*E*A)*(-x**4 + 4*L**3*x)

plt.plot(x, u_q, label='generalized c')
plt.plot(x, u_L, label='lagrange', linestyle=':')
plt.plot(x, u_analytic, label='analytic')
plt.legend()
plt.grid()
plt.xlabel('x / m')
plt.ylabel('u / m')
plt.savefig('u_s.pdf')
```
